```python
import pandas as pd
```

```python
df["Age"] = pd.to_numeric(df["Age"], errors="coerce")
```

```python
df["Age"] = df["Age"].fillna(df["Age"].mean()).round().astype(int)
df["Marks"] = df["Marks"].fillna(df["Marks"].mean()).round(2)
df["Attendance"] = df["Attendance"].fillna(df["Attendance"].mean()).round(2)
```

```python
df["Course"] = df["Course"].fillna(df["Course"].mode()[0])
```

```python
df = df.drop_duplicates()
```

```python
print("AFTER CLEANING: Cleaned dataset")
display(df)
```

```python
print("Missing values after cleaning:")
print(df.isnull().sum())
```

```python
print("Duplicate records after cleaning:")
print(df.duplicated().sum())

print("Data types after cleaning:")
print(df.dtypes)

print("Cleaned dataset shape:")
print(df.shape)

print("Basic statistics after cleaning:")
display(df.describe())
```
